```python
def convert_sqft(x):
    x = str(x)
    if '-' in x:
        a, b = x.split('-')
        return (float(a) + float(b)) / 2
    try:
        return float(x)
    except:
        return None                                # for nan
```

```python
df['total_sqft'] = df['total_sqft'].apply(convert_sqft)                     # applying fun on df['total_sqft'] for every row value
```

```python
def remove_outliers(df):
    df_out = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        mean = subdf['price_per_sqft'].mean()
        std = subdf['price_per_sqft'].std()
        reduced = subdf[
            (subdf['price_per_sqft'] > mean - std) &
            (subdf['price_per_sqft'] < mean + std)
        ]
        df_out = pd.concat([df_out, reduced], ignore_index=True)
    return df_out
```

```python
df = remove_outliers(df)
```

```python
df.describe()
```

```python
def remove_bhk_outliers(df):
    df_out = pd.DataFrame()

    for loc, loc_df in df.groupby('location'):
        bhk_stats = {}

        # calculate mean price_per_sqft for each BHK
        for bhk, bhk_df in loc_df.groupby('bhk'):
            bhk_stats[bhk] = bhk_df['price_per_sqft'].mean()

        temp_df = pd.DataFrame()

        for bhk, bhk_df in loc_df.groupby('bhk'):
            if bhk - 1 in bhk_stats:
                # compare with previous BHK
                if bhk_stats[bhk] < bhk_stats[bhk - 1]:
                    continue  # remove this BHK group
            temp_df = pd.concat([temp_df, bhk_df])

        df_out = pd.concat([df_out, temp_df])

    return df_out
```

```python
df = remove_bhk_outliers(df)
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('price', axis=1)
y = df['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

y_pred = model.predict(X_test)

print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

```python
from sklearn.linear_model import Ridge

ridge_regression = Ridge(alpha = 1.0)                    # the 'Lamda' hyperparameter is alpha here.
ridge_regression.fit(X_train, y_train)
```

```python
ridge_pred = ridge_regression.predict(X_test)

print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, ridge_pred))
```

```python
from sklearn.linear_model import Lasso

lasso_regression = Lasso(alpha = 1.0)                    # the 'Lamda' hyperparameter is alpha here.
lasso_regression.fit(X_train, y_train)
```

```python
lasso_pred = lasso_regression.predict(X_test)

print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, lasso_pred))
```
